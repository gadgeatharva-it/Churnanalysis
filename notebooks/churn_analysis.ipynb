# ChurnPredict AI - Customer Churn Analysis

This notebook walks through the core data science workflow for the IBM Telco Customer Churn dataset: loading, cleaning, exploratory analysis, model training, evaluation, and business interpretation.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.append(str(PROJECT_ROOT / 'src'))

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from data_preprocessing import load_dataset, clean_dataset, dataset_quality_report, churn_rate_by, calculate_kpis
from train_model import train_and_select_best_model

sns.set_theme(style='whitegrid')
```

## 1. Load and Inspect Data

```python
raw_df = load_dataset(PROJECT_ROOT / 'data' / 'customer_churn.csv')
raw_df.head()
```

```python
report = dataset_quality_report(raw_df)
report['shape'], report['duplicate_rows'], pd.Series(report['missing_values']).sort_values(ascending=False).head()
```

## 2. Data Cleaning

```python
df = clean_dataset(raw_df)
df.info()
```

## 3. KPI Summary

```python
calculate_kpis(df)
```

## 4. Exploratory Data Analysis

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
sns.countplot(data=df, x='Churn', ax=axes[0, 0])
axes[0, 0].set_title('Overall Churn Distribution')
sns.countplot(data=df, x='Contract', hue='Churn', ax=axes[0, 1])
axes[0, 1].set_title('Churn by Contract Type')
sns.histplot(data=df, x='tenure', hue='Churn', multiple='stack', ax=axes[1, 0])
axes[1, 0].set_title('Tenure and Churn')
sns.boxplot(data=df, x='Churn', y='MonthlyCharges', ax=axes[1, 1])
axes[1, 1].set_title('Monthly Charges and Churn')
plt.tight_layout()
```

```python
for column in ['gender', 'Contract', 'InternetService', 'PaymentMethod']:
    display(churn_rate_by(df, column))
```

```python
numeric_corr = df[['SeniorCitizen', 'tenure', 'MonthlyCharges', 'TotalCharges']].corr()
plt.figure(figsize=(8, 5))
sns.heatmap(numeric_corr, annot=True, cmap='RdBu_r', center=0)
plt.title('Correlation Heatmap for Numerical Features')
plt.show()
```
